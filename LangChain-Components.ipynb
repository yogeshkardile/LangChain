## Models

Models are the core interfaces through which you interact with AI models. LangChain provides a common interface for interacting with different AI model providers.

### Sample Code

First, install the required package:

```python
%pip install -U langchain-openai
```

Then import and initialize the model:

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

response = model.invoke("What is Machine Learning?")

print(response.content)
```

### Explanation

- **`ChatOpenAI`** → Provides a LangChain interface for interacting with OpenAI chat models.
- **`model`** → Creates an instance of the AI model.
- **`temperature=0`** → Makes the model's output more deterministic and consistent.
- **`invoke()`** → Sends a prompt to the model and returns the model's response.
- **`response.content`** → Extracts the actual text generated by the model.

### Flow

```text
Prompt
  ↓
LangChain Model Interface
  ↓
OpenAI Chat Model
  ↓
Response
```

> **Note:** You need to configure an OpenAI API key before running the model.

## Prompts

Prompts are instructions or inputs given to an AI model to guide its response. In LangChain, the **Prompt Templates** component helps create reusable and dynamic prompts by using variables.

### Sample Code

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple words."
)

formatted_prompt = prompt.invoke({
    "topic": "Machine Learning"
})

print(formatted_prompt.text)
```

### Output

```text
Explain Machine Learning in simple words.
```

### Explanation

- **`PromptTemplate`** → Used to create reusable prompt templates.
- **`input_variables`** → Defines the variables that will be dynamically inserted into the prompt.
- **`template`** → Defines the structure of the prompt.
- **`invoke()`** → Provides values for the defined variables and creates the final prompt.
- **`formatted_prompt.text`** → Returns the generated prompt as text.

### Why Use Prompt Templates?

Prompt templates are useful because they allow us to:

- Create **reusable prompts**
- Dynamically insert user input
- Maintain **consistent prompt structure**
- Easily modify prompts without rewriting the entire code

### Flow

```text
Template
   ↓
Input Variables
   ↓
Formatted Prompt
   ↓
AI Model
   ↓
Response
```

## Chains

Chains allow us to **connect multiple components or operations together** to create a complete workflow. The output of one component can be passed as the input to the next component.

### Sample Code

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate

# Create the model
model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

# Create the prompt
prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple words."
)

# Create a chain
chain = prompt | model

# Run the chain
response = chain.invoke({
    "topic": "Machine Learning"
})

print(response.content)
```

### Explanation

- **`PromptTemplate`** → Creates a reusable prompt.
- **`ChatOpenAI`** → Represents the AI model.
- **`|` (pipe operator)** → Connects the components together.
- **`chain`** → Combines the prompt and model into a single workflow.
- **`invoke()`** → Executes the complete chain.
- **`response.content`** → Extracts the model's generated response.

### Flow

```text
User Input
    ↓
Prompt Template
    ↓
AI Model
    ↓
Response
```

### Why Use Chains?

Chains are useful because they allow us to:

- **Connect multiple components** into a single workflow.
- **Reuse workflows** without manually calling each component.
- **Pass outputs between components** automatically.
- Build more complex **LLM applications** by combining prompts, models, parsers, retrievers, and other components.

## Memory

Memory allows an AI application to **retain information from previous interactions** and use that information in future interactions. This helps create conversational applications where the model can maintain context across multiple messages.

### Example

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, AIMessage

model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

messages = [
    HumanMessage(content="My name is Yogesh."),
    AIMessage(content="Nice to meet you, Yogesh!"),
    HumanMessage(content="What is my name?")
]

response = model.invoke(messages)

print(response.content)
```

### Explanation

- **`HumanMessage`** → Represents a message sent by the user.
- **`AIMessage`** → Represents a previous response from the AI.
- **`messages`** → Stores the conversation history.
- **`model.invoke()`** → Sends the conversation history to the model.
- The model can use the previous messages to **maintain context**.

### Important Note

Modern LangChain applications generally manage conversation history using **message history or memory integrations**, rather than relying on the older `ConversationBufferMemory` approach.

### Flow

```text
User Message
     ↓
Conversation History
     ↓
AI Model
     ↓
AI Response
     ↓
Conversation History Updated
```

### Why Use Memory?

Memory is useful for:

- Maintaining **conversation context**
- Creating **chatbots with multi-turn conversations**
- Remembering information provided earlier in a conversation
- Building more **personalized AI applications**

## Indexes

Indexes are data structures used to **organize and efficiently retrieve information** from external data sources such as documents, PDFs, websites, or databases.

In LangChain, indexes are commonly used in **Retrieval-Augmented Generation (RAG)** applications. They help convert documents into searchable representations so that relevant information can be retrieved when a user asks a question.

### Basic RAG Flow

```text
Documents
    ↓
Document Loader
    ↓
Text Splitter
    ↓
Embeddings
    ↓
Vector Database
    ↓
Retriever
    ↓
Relevant Documents
    ↓
LLM
    ↓
Answer
```

### Sample Code

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS

# Load document
loader = TextLoader("example.txt")
documents = loader.load()

# Split document into chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(documents)

# Create embeddings
embeddings = OpenAIEmbeddings()

# Create vector index
vectorstore = FAISS.from_documents(
    chunks,
    embeddings
)

# Search for relevant information
results = vectorstore.similarity_search(
    "What is machine learning?"
)

for doc in results:
    print(doc.page_content)
```

### Explanation

- **Document Loader** → Loads data from sources such as PDFs, text files, websites, etc.
- **Text Splitter** → Breaks large documents into smaller chunks.
- **Embeddings** → Converts text into numerical vectors that represent its meaning.
- **Vector Store / Index** → Stores and indexes the vectors for efficient similarity search.
- **Similarity Search** → Finds documents that are semantically similar to the user's query.

### Why Use Indexes?

Indexes are useful because they allow us to:

- Efficiently **search large amounts of information**
- Retrieve **relevant documents** based on a query
- Build **RAG applications**
- Connect LLMs with **external knowledge sources**
- Avoid sending entire documents to the LLM for every query

### Simple Example

Suppose we have 1,000 PDF documents.

Without an index:

```text
User Query
    ↓
Search through all documents
    ↓
Find relevant information
```

With an index:

```text
User Query
    ↓
Convert Query → Embedding
    ↓
Search Index
    ↓
Retrieve Relevant Chunks
    ↓
LLM
    ↓
Answer
```

> **Note:** In modern LangChain terminology, the indexing/retrieval workflow is typically implemented using **document loaders, text splitters, embeddings, vector stores, and retrievers** rather than treating "Index" as a single standalone component.

## Agents

Agents are systems that allow an LLM to **decide which actions or tools to use** to accomplish a given task.

Unlike a simple chain, where the execution flow is predefined, an agent can **dynamically decide what to do next** based on the user's request and the available tools.

### Sample Code

```python
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

def calculator(expression: str) -> str:
    """Calculate a mathematical expression."""
    return str(eval(expression))

agent = create_agent(
    model=model,
    tools=[calculator]
)

response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is 25 * 40?"
        }
    ]
})

print(response["messages"][-1].content)
```

### Explanation

- **`model`** → The LLM responsible for reasoning and deciding what action to take.
- **`tools`** → Functions or external capabilities that the agent can use.
- **`create_agent()`** → Creates an agent that can decide when and how to use the available tools.
- **`invoke()`** → Sends the user's task to the agent.
- The agent can **select the appropriate tool**, execute it, and use the result to produce the final answer.

### Chain vs Agent

**Chain:**

```text
Input
  ↓
Prompt
  ↓
Model
  ↓
Output
```

The workflow is **predefined**.

**Agent:**

```text
Input
  ↓
LLM
  ↓
Decide which tool to use
  ↓
Tool
  ↓
Observe result
  ↓
LLM
  ↓
Final Answer
```

The workflow is **dynamic**.

### Why Use Agents?

Agents are useful for:

- Using **multiple tools**
- Performing **multi-step tasks**
- Making decisions dynamically
- Interacting with **APIs, databases, search engines, and other tools**
- Building autonomous AI applications

> **Key Idea:** A chain follows a predefined workflow, while an agent allows the LLM to decide the next action based on the task.
